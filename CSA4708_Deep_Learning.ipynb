```python
# 1.Vector and Matrix Operations for Deep Learning
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Vector Addition:", a + b)
print("Vector Subtraction:", a - b)
print("Vector Multiplication:", a * b)
print("Dot Product:", np.dot(a, b))

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Matrix Addition:\n", A + B)
print("Matrix Multiplication:\n", np.dot(A, B))
print("Transpose:\n", A.T)
```

```python
#2: Covariance, Eigenvalues and Eigenvectors Analysis
import numpy as np

data = np.array([[2, 4, 6],
                 [3, 5, 7],
                 [4, 6, 8]])
cov = np.cov(data)
print("Covariance Matrix:\n", cov)

A = np.array([[4, 1],
              [1, 3]])
values, vectors = np.linalg.eig(A)

print("Eigenvalues:", values)
print("Eigenvectors:\n", vectors)
```

```python
#3: Probability Distributions and Bayes' Theorem
from scipy.stats import binom

# Binomial probability
p = binom.pmf(3, 5, 0.5)
print("Binomial Probability:", round(p, 4))

# Bayes' Theorem
P_A = 0.01
P_B_given_A = 0.90
P_B_given_not_A = 0.05

P_not_A = 1 - P_A

P_A_given_B = (P_B_given_A * P_A) / (
    P_B_given_A * P_A +
        P_B_given_not_A * P_not_A
        )

print("Bayes Probability:", round(P_A_given_B, 4))
```

```python
#4: Entropy, Cross-Entropy and KL Divergence
import numpy as np
from scipy.stats import entropy

p = np.array([0.7, 0.3])
q = np.array([0.6, 0.4])

H = entropy(p, base=2)
CE = -np.sum(p * np.log2(q))
KL = entropy(p, q, base=2)

print("Entropy:", round(H, 4))
print("Cross Entropy:", round(CE, 4))
print("KL Divergence:", round(KL, 4))
```

```python
#5: Bias-Variance, Overfitting and Underfitting Analysis
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

X, y = make_classification(
    n_samples=200,
    n_features=5,
    random_state=42
)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

model = DecisionTreeClassifier(max_depth=2, random_state=42)
model.fit(X_train, y_train)

train_pred = model.predict(X_train)
test_pred = model.predict(X_test)

print("Training Accuracy:", accuracy_score(y_train, train_pred))
print("Testing Accuracy:", accuracy_score(y_test, test_pred))
```
